## 题目 3：Softmax 损失函数 (Softmax Loss)

在 `src/simple_ml.py` 中的 `softmax_loss()` 函数里实现 Softmax（又称交叉熵 / Cross-Entropy）损失。复习一下（希望这是已掌握的内容，我们也会在 9/1 的课程中讲授）：对于取值范围为 $y \in \{1,\ldots,k\}$ 的多分类输出，Softmax 损失接收一个 Logits 向量 $z \in \mathbb{R}^k$ 和真实类别 $y \in \{1,\ldots,k\}$，返回的损失定义为：
$$
\ell_{\mathrm{softmax}}(z, y) = \log\sum_{i=1}^k \exp z_i - z_y.
$$

请注意，正如其 docstring 所述，`softmax_loss()` 接收一个 Logits 的**二维数组**（即整个 Batch 中不同样本的 $k$ 维预测分值），以及对应的真实标签一维数组，并且应该输出整个 Batch 的**平均** Softmax 损失。为了高效正确地实现它，你**不应该**使用任何循环，而应该完全使用 NumPy 原生的向量化操作进行计算（供参考：我们的参考解决方案只有短短一行代码）。

注意：在真实的工业级 Softmax 损失实现中，通常需要对 Logits 进行平移缩放以防止数值上溢（Numerical Overflow），但在这里无需担心这一点（即使不处理数值平移，后续作业也能正常运行）。下方代码用于运行测试用例。

## 题目 4：Softmax 回归的随机梯度下降 (SGD)

在本题中，你将为（线性）Softmax 回归实现随机梯度下降（Stochastic Gradient Descent, SGD）。换言之，正如我们在 9/1 课堂上讨论的那样，我们考虑一个假设函数（Hypothesis Function），它通过如下方式将 $n$ 维输入映射为 $k$ 维 Logits：
$$
h(x) = \Theta^T x
$$
其中 $x \in \mathbb{R}^n$ 为输入向量，$\Theta \in \mathbb{R}^{n \times k}$ 为模型参数矩阵。给定数据集 $\{(x^{(i)} \in \mathbb{R}^n, y^{(i)} \in \{1,\ldots,k\})\}_{i=1}^m$，Softmax 回归对应的优化目标为：
$$
\operatorname*{minimize}_{\Theta} \; \frac{1}{m} \sum_{i=1}^m \ell_{\mathrm{softmax}}(\Theta^T x^{(i)}, y^{(i)}).
$$

回忆课堂内容，线性 Softmax 目标的梯度公式为：
$$
\nabla_\Theta \ell_{\mathrm{softmax}}(\Theta^T x, y) = x (z - e_y)^T
$$
其中：
$$
z = \frac{\exp(\Theta^T x)}{1^T \exp(\Theta^T x)} \equiv \operatorname*{normalize}(\exp(\Theta^T x))
$$
（即 $z$ 是经过归一化的 Softmax 概率向量），$e_y$ 表示第 $y$ 个单位基向量（即在第 $y$ 个位置为 1、其余位置全为 0 的 One-Hot 向量）。

我们也可以用课堂上讨论过的更紧凑的矩阵记号来表示。即：若令 $X \in \mathbb{R}^{m \times n}$ 表示由 $m$ 个样本组成的设计矩阵（可以是整个数据集或一个 Minibatch），$y \in \{1,\ldots,k\}^m$ 表示对应的标签向量，并用 $\ell_{\mathrm{softmax}}$ 表示平均 Softmax 损失，则有：
$$
\nabla_\Theta \ell_{\mathrm{softmax}}(X \Theta, y) = \frac{1}{m} X^T (Z - I_y)
$$
其中：
$$
Z = \operatorname*{normalize}(\exp(X \Theta)) \quad \text{（按行应用归一化）}
$$
表示归一化后的概率矩阵，$I_y \in \mathbb{R}^{m \times k}$ 表示由标签 $y$ 的 One-Hot 向量按行拼接而成的矩阵。

利用上述梯度公式，在 `src/simple_ml.py` 中实现 `softmax_regression_epoch()` 函数。该函数使用给定的学习率 `lr` 和 Minibatch 大小 `batch` 执行一个完整的 SGD 训练周期（Epoch，即遍历数据集一遍）。正如文档注释所要求的，你的函数应当**原地（in-place）修改**参数矩阵 `theta`。实现后，运行本地测试。

## 题目 5：两层神经网络的随机梯度下降 (SGD)

完成了线性分类器的 SGD 之后，现在我们考虑一个简单的两层神经网络。具体而言，对于输入 $x \in \mathbb{R}^n$，我们考虑如下形式的两层神经网络（不含偏置项 Bias）：
$$
z = W_2^T \mathrm{ReLU}(W_1^T x)
$$
其中 $W_1 \in \mathbb{R}^{n \times d}$ 与 $W_2 \in \mathbb{R}^{d \times k}$ 分别表示网络的权重矩阵（网络包含 $d$ 维隐藏单元），$z \in \mathbb{R}^k$ 表示网络输出的 Logits。我们同样使用 Softmax / 交叉熵损失，即需要求解如下优化问题：
$$
\operatorname*{minimize}_{W_1, W_2} \;\; \frac{1}{m} \sum_{i=1}^m \ell_{\mathrm{softmax}}(W_2^T \mathrm{ReLU}(W_1^T x^{(i)}), y^{(i)}).
$$
或者，用批量矩阵 $X \in \mathbb{R}^{m \times n}$ 形式可以写为：
$$
\operatorname*{minimize}_{W_1, W_2} \;\; \ell_{\mathrm{softmax}}(\mathrm{ReLU}(X W_1) W_2, y).
$$

利用链式法则（Chain Rule），我们可以推导出该网络的反向传播更新公式（我们将在 9/8 的课堂上简要介绍推导过程，这里直接给出最终公式以便于实现）：
$$
\begin{split}
Z_1 \in \mathbb{R}^{m \times d} & = \mathrm{ReLU}(X W_1) \\
G_2 \in \mathbb{R}^{m \times k} & = \operatorname*{normalize}(\exp(Z_1 W_2)) - I_y \\
G_1 \in \mathbb{R}^{m \times d} & = \mathrm{1}\{Z_1 > 0\} \circ (G_2 W_2^T)
\end{split}
$$
其中 $\mathrm{1}\{Z_1 > 0\}$ 为二值掩码矩阵，其元素取值为 1 或 0，取决于 $Z_1$ 中的对应项是否严格大于 0；符号 $\circ$ 表示逐元素相乘（Element-wise Multiplication）。由此，目标函数关于各权重矩阵的梯度为：
$$
\begin{split}
\nabla_{W_1} \ell_{\mathrm{softmax}}(\mathrm{ReLU}(X W_1) W_2, y) & = \frac{1}{m} X^T G_1  \\
\nabla_{W_2} \ell_{\mathrm{softmax}}(\mathrm{ReLU}(X W_1) W_2, y) & = \frac{1}{m} Z_1^T G_2.  \\
\end{split}
$$

**注意**：如果你在 9/8 课程之前觉得上述公式细节稍显晦涩，不必过于担心。这正是标准两层 ReLU 网络的反向传播公式：$Z_1$ 项计算的是“前向（Forward）”过程，而 $G_2$ 和 $G_1$ 则代表“反向（Backward）”过程。公式的具体形式可能会因不同文献中神经网络的转置记号、损失函数的具体表述形式等有所差异。如果你对深层网络有些许印象，并在听完 9/8 的讲授后豁然开朗，那么这在背景知识层面上就已经完全足够了（毕竟深度学习系统的核心目的之一，就是在很大程度上将我们从这些手工计算中解放出来）。但如果你觉得这些概念完全是天方夜谭，那么建议在学习本课之前先补充机器学习与神经网络的基础知识。

利用这些梯度公式，在 `src/simple_ml.py` 中编写 `nn_epoch()` 函数。与前一题类似，你的实现应当原地修改 `W1` 和 `W2` 数组。实现后运行下方测试。请务必使用上述公式所指示的矩阵运算来实现函数：这比尝试使用循环要快得多、高效得多，而且代码量也大幅精简。

让我们剖析一下该函数的参数。这个函数在逻辑上基本是 Python 实现的镜像映射，但需要传入额外的参数，因为我们在这里直接操作的是底层数组数据的**原始裸指针（Raw Pointers）**，而非任何高级的“矩阵”数据结构。具体而言，`X`、`y` 和 `theta` 分别是指向上一节 NumPy 数组底层原始连续内存数据的指针；对于二维数组，它们以 C 风格（**行优先 / Row-major**）格式存储，意味着 $X$ 的第一行作为起始字节连续存放，接着是第二行，依此类推（这与按列优先 / Column-major 存储形成对比）。我们还假定数据中没有内存填充（Padding），即第二行紧随第一行末尾，中间没有任何补位对齐字节（这些问题会在课程后续讨论，此处暂不涉及）。因为只传入了原始裸指针，为了获知底层矩阵的实际几何尺寸，我们必须显式将尺寸传入函数，这正是参数 `m`、`n` 和 `k` 的作用。

关于如何寻址数据的示例：因为 `X` 表示一个按行优先存储的 $m \times n$ 矩阵，如果我们想要访问 $X$ 的第 $(i, j)$ 个元素（即第 $i$ 行、第 $j$ 列的元素），我们将使用如下一维索引公式：
```cpp
X[i*n + j]
```
即：因为 `X` 有 $n$ 列且各列顺序存放，我们需要跳过 `i*n` 个元素才能到达第 $i$ 行的开头；如果再取这一行中的第 $j$ 个元素，就得到了上述表达式。同样的逻辑也适用于 `theta` 矩阵，但关键在于：因为 `theta` 是一个 $n \times k$ 的矩阵，要访问其第 $(i, j)$ 个元素，你需要使用索引：
```cpp
theta[i*k + j]
```
与 Python 不同，在 C++ 中直接像这样访问底层内存时必须极其小心，并逐渐习惯这种指针偏移记法（或者构建辅助数据结构来更直观地访问，但对于本作业，你只需要坚持使用裸指针偏移即可）。

实现的第二部分是实际提供 Python 接口的 pybind11 胶水代码：
